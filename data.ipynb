```python
match_df = df.groupby('match_id').agg({
    'batting_team': 'first',
    'bowling_team': 'first',
    'venue': 'first',
    'city': 'first',
    'toss_winner': 'first',
    'toss_decision': 'first',
    'year': 'first',
    'month': 'first',
    'day': 'first',
    'match_won_by': 'first'
}).reset_index()
```

```python
TEAM_MAP = {
    "mumbai indians": "mi",
    "chennai super kings": "csk",
    "royal challengers bangalore": "rcb",
    "royal challengers bengaluru": "rcb",
    "sunrisers hyderabad": "srh",
    "kolkata knight riders": "kkr",
    "delhi capitals": "dc",
    "delhi daredevils": "dc",
    "rajasthan royals": "rr",
    "punjab kings": "pbks",
    "kings xi punjab": "pbks",
    "lucknow super giants": "lsg",
    "gujarat titans": "gt",
    "rising pune supergiant": "rps",
    "rising pune supergiants": "rps",
    "gujarat lions": "gl",
    "kochi tuskers kerala": "ktk",
    "deccan chargers": "dch",
    "pune warriors": "pwi",
}

HOME_VENUES = {
    "mi":   ["wankhede stadium", "brabourne stadium"],
    "csk":  ["ma chidambaram stadium", "chepauk"],
    "rcb":  ["m chinnaswamy stadium", "chinnaswamy"],
    "srh":  ["rajiv gandhi international stadium", "uppal"],
    "kkr":  ["eden gardens"],
    "dc":   ["arun jaitley stadium", "feroz shah kotla"],
    "rr":   ["sawai mansingh stadium", "sms stadium"],
    "pbks": ["punjab cricket association stadium", "is bindra stadium", "mohali"],
    "lsg":  ["ekana cricket stadium", "ekana"],
    "gt":   ["narendra modi stadium", "motera"],
}
```

```python
def map_team(team):
    if pd.isna(team):
        return "unknown"
    return TEAM_MAP.get(str(team).lower().strip(), str(team).lower().strip())

def is_home(team, venue):
    team = map_team(team)
    venue = str(venue).lower().strip()
    home_list = HOME_VENUES.get(team, [])
    return int(any(h in venue for h in home_list))
```

```python
for col in ['batting_team','bowling_team','toss_winner','match_won_by']:
    match_df[col] = match_df[col].apply(map_team)
```

```python
match_df = match_df[match_df['match_won_by'] != 'unknown']
```

```python
team_counts = match_df['match_won_by'].value_counts()
valid_teams = team_counts[team_counts > 30].index
match_df = match_df[match_df['match_won_by'].isin(valid_teams)]
```

```python
match_df = match_df.sort_values(by=['year','month','day'])
```

```python
team_stats = {}

bat_strength = []
bowl_strength = []
```

```python
for _, row in match_df.iterrows():
    bt = row['batting_team']
    bw = row['bowling_team']
    winner = row['match_won_by']
    
    for t in [bt, bw]:
        if t not in team_stats:
            team_stats[t] = {'matches': 0, 'wins': 0}
    
    bt_wr = team_stats[bt]['wins'] / team_stats[bt]['matches'] if team_stats[bt]['matches'] > 0 else 0.5
    bw_wr = team_stats[bw]['wins'] / team_stats[bw]['matches'] if team_stats[bw]['matches'] > 0 else 0.5
    
    bat_strength.append(bt_wr)
    bowl_strength.append(bw_wr)
    
    team_stats[bt]['matches'] += 1
    team_stats[bw]['matches'] += 1
    team_stats[winner]['wins'] += 1

match_df['batting_strength'] = bat_strength
match_df['bowling_strength'] = bowl_strength
```

```python
match_df['batting_home'] = match_df.apply(
    lambda x: is_home(x['batting_team'], x['venue']), axis=1)

match_df['bowling_home'] = match_df.apply(
    lambda x: is_home(x['bowling_team'], x['venue']), axis=1)
```

```python
X = match_df.drop(columns=['match_id','match_won_by'])
y = match_df['match_won_by']
```

```python
from sklearn.preprocessing import LabelEncoder, OneHotEncoder
```

```python
le = LabelEncoder()
y_encoded = le.fit_transform(y)
```

```python
categorical_cols = X.select_dtypes(include=['object']).columns.tolist()
numerical_cols = X.select_dtypes(include=['int64','float64']).columns.tolist()
```

```python
from sklearn.compose import ColumnTransformer
```

```python
preprocessor = ColumnTransformer([
    ('cat', OneHotEncoder(handle_unknown='ignore'), categorical_cols),
    ('num', 'passthrough', numerical_cols)
])
```

# MODELS

```python
from sklearn.ensemble import RandomForestClassifier, VotingClassifier
from xgboost import XGBClassifier
from lightgbm import LGBMClassifier
```

```python
rf = RandomForestClassifier(n_estimators=200, random_state=42)
```

```python
xgb = XGBClassifier(eval_metric='mlogloss', use_label_encoder=False)
```

```python
lgbm = LGBMClassifier()
```

```python
ensemble = VotingClassifier(
    estimators=[('rf', rf), ('xgb', xgb), ('lgbm', lgbm)],
    voting='soft'
)
```

```python
from sklearn.pipeline import Pipeline
```

```python
pipeline = Pipeline([
    ('preprocessor', preprocessor),
    ('model', ensemble)
])
```

```python
from sklearn.model_selection import TimeSeriesSplit
import numpy as np
```

```python
tscv = TimeSeriesSplit(n_splits=5)

y_pred = np.zeros(len(y_encoded))

for train_idx, test_idx in tscv.split(X):
    X_train, X_test = X.iloc[train_idx], X.iloc[test_idx]
    y_train = y_encoded[train_idx]

    pipeline.fit(X_train, y_train)
    y_pred[test_idx] = pipeline.predict(X_test)

y_pred = y_pred.astype(int)
```

```python
valid_idx = y_pred != 0
```

```python
from sklearn.metrics import accuracy_score, classification_report
```

```python
print("Accuracy:", accuracy_score(y_encoded[valid_idx], y_pred[valid_idx]))
print("\nClassification Report:\n")
print(classification_report(y_encoded[valid_idx], y_pred[valid_idx], target_names=le.classes_))
```
